```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("amirmotefaker/supply-chain-dataset")

print("Path to dataset files:", path)
```

```python
import os 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
csv_file = os.path.join(path,os.listdir(path)[0])
df = pd.read_csv(csv_file)

df.head(5)
df.sample(5)
```

```python
num_col = df.select_dtypes(include=['int64','float64'])
num_col.corr()
sns.heatmap(num_col.corr())
```

```python
df.hist(figsize=(14,12))
```

```python
outliers = []
for col in num_col:
        q1 = df[col].quantile(.25)
        q3 = df[col].quantile(.75)
        iqr = q3 - q1
        lower_limit = q1 - 1.5*iqr
        upper_limit = q3 + 1.5*iqr
        count = ((df[col]>upper_limit) | (df[col]<lower_limit)).sum()
        outliers.append([col,count])

outlier_df = pd.DataFrame(outliers,columns = ['column','number_of_out'])

outlier_df
```

```python
#  which category of produccts corressponds to the most revenue
df.groupby('Product type')['Revenue generated'].sum().sort_values(ascending=False)

df.groupby('Product type')['Revenue generated'].sum().plot(kind = 'bar')
```

```python
#  which category of people spend more
df.groupby('Customer demographics')['Revenue generated'].sum().sort_values(ascending=False)

df.groupby('Customer demographics')['Revenue generated'].sum().plot(kind = 'bar')
```

```python
#  which suppllier gives good products
df.groupby('Supplier name')['Defect rates'].mean().sort_values()
```

```python
# fastest suppliers needed for fast supply to customers
df.groupby('Supplier name')['Lead times'].mean().sort_values()
```

```python
# cheapest modes for tranport
df.groupby('Transportation modes')['Shipping costs'].mean().sort_values()
```

```python
df.groupby('Transportation modes')['Shipping times'].mean().sort_values()
```

```python
df['Revenue_per_Unit'] = (
    df['Revenue generated']
    / df['Number of products sold']
)
df.groupby('Product type')['Revenue_per_Unit'].mean().sort_values()
```

```python
df.groupby('Product type')['Defect rates'].mean()
```

```python
df['Revenue_Cost_Ratio'] = (
    df['Revenue generated']
    / df['Manufacturing costs']
)

df.groupby('Product type')['Revenue_Cost_Ratio'] \
  .mean() \
  .sort_values(ascending=False)
```

```python
df.groupby('SKU')['Revenue_Cost_Ratio'] \
  .mean() \
  .sort_values(ascending=False).head(5)
```

```python
df['Profit'] = (
    df['Revenue generated']
    - df['Manufacturing costs']
)
df.groupby('Product type')['Profit'] \
  .sum() \
  .sort_values(ascending=False)
```

```python
df.groupby('SKU')['Profit'] \
  .sum() \
  .sort_values(ascending=False).head(5)
```
